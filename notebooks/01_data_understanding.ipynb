```python
import os
import pandas as pd

folder = "../data/raw"
files = os.listdir(folder)

total_rows = 0
non_missing_totals = None

for file in files:
    if not file.endswith(".csv"):
        continue

    file_path = os.path.join(folder, file)
    current_df = pd.read_csv(file_path)

    current_counts = current_df.notna().sum()

    if non_missing_totals is None:
        non_missing_totals = current_counts
    else:
        non_missing_totals = non_missing_totals.add(
            current_counts,
            fill_value=0
        )

    total_rows += len(current_df)
```

```python
availability_percent = (non_missing_totals / total_rows) * 100
availability_percent = availability_percent.sort_values(ascending=False)
availability_percent = availability_percent[availability_percent > 80]

availability_percent
```
