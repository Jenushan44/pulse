```python
import pandas as pd

training_df = pd.read_csv("../data/processed/training_data.csv")

training_df["date"] = pd.to_datetime(training_df["date"])

training_df = training_df.sort_values(['serial_number', 'date']).reset_index(drop=True)
training_df["smart_5_raw_change_1d"] = training_df.groupby("serial_number")['smart_5_raw'].diff()
training_df["smart_5_raw_change_7d"] = training_df.groupby("serial_number")['smart_5_raw'].diff(periods=7)
training_df["smart_5_raw_change_30d"] = training_df.groupby("serial_number")['smart_5_raw'].diff(periods=30)

training_df["smart_5_raw_rolling_mean_7d"] = training_df.groupby("serial_number")["smart_5_raw"].rolling(window=7, min_periods=7).mean().reset_index(level=0, drop=True)
training_df["smart_5_raw_rolling_std_7d"] = training_df.groupby("serial_number")["smart_5_raw"].rolling(window=7, min_periods=7).std().reset_index(level=0, drop=True)

training_df["smart_5_raw_rolling_mean_30d"] = training_df.groupby("serial_number")["smart_5_raw"].rolling(window=30, min_periods=30).mean().reset_index(level=0, drop=True)

print(training_df.loc[training_df['serial_number'] == '93M0A03MV4MJ', ["serial_number", "date", "smart_5_raw", "smart_5_raw_change_1d", "smart_5_raw_change_7d", "smart_5_raw_change_30d", "smart_5_raw_rolling_mean_7d", "smart_5_raw_rolling_std_7d", "smart_5_raw_rolling_mean_30d"]])
```
