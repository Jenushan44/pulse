# Class Imbalance Analysis

Analyze the distribution of the `failure_within_30_days` target to check whether or not the training dataset is imbalanced and examine how the failure examples are distributed across the drive models.

```python
import pandas as pd

training_df = pd.read_csv("../data/processed/training_data.csv")

fail_count = len(training_df[training_df['failure_within_30_days'] == 1])
non_fail_count = len(training_df[training_df['failure_within_30_days'] == 0])
total_count = fail_count + non_fail_count

failure_percentage = (fail_count / total_count) * 100
non_failure_percentage = (non_fail_count / total_count) * 100


print(fail_count)
print(non_fail_count)

print(failure_percentage)
print(non_failure_percentage)
```

## Failures by Drive Model

Examine how the positive failure examples are distributed across different drive models to determine whether or not certain models make up a larger part of the failure data.

```python
failure_rows = training_df[training_df['failure_within_30_days'] == 1]
drive_models = failure_rows['model']

print(drive_models.value_counts())
```

```python
total_observations = (training_df['model']).value_counts()
positive_observations = drive_models.value_counts()

positive_rate_per_model = (positive_observations / total_observations) * 100
positive_rate_per_model = positive_rate_per_model.fillna(0).sort_values(ascending=False)

print(positive_rate_per_model)
```

## Conclusion

The training dataset contains 22,786 positive observations and 79,337 negative observations. About 22.3% of the observations are positive and 77.7% are negative, showing that the dataset has a class imbalance. The positive observations are also not distributed evenly across the drive models. Some models have much higher positive-label rates than others and some models have no positive observations in the training dataset. Since the majority class makes up about 77.7% of the dataset, the accuracy may be misleading when evaluating the machine learning model. A model that keeps predicting the majority class could have high accuracy while also still failing to identify drives that are approaching failure.
